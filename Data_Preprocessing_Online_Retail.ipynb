Memeriksa bagian dataset yang tersedia dari UCI


```python
print("Kolom fitur:")
print(online_retail.data.features.columns.tolist())

print("\nKolom ID:")
if online_retail.data.ids is not None:
    print(online_retail.data.ids.columns.tolist())
    display(online_retail.data.ids.head())
else:
    print("Tidak ada kolom ID terpisah.")

print("\nData asli tersedia:")
print(online_retail.data.original is not None)
```

Menggabungkan kolom fitur dan ID

```python
df_raw = pd.concat(
    [
        online_retail.data.features,
        online_retail.data.ids
    ],
    axis=1
).copy()
```

Membuat data kerja dari data mentah lengkap

```python
df = df_raw.copy()

print("Ukuran dataset lengkap:", df.shape)
print("\nDaftar kolom:")
print(df.columns.tolist())

print("\nLima baris pertama:")
display(df.head())
```

**2. Data Preprocessing**

Pemeriksaan kualitas dataset lengkap

```python
print("Ukuran data:", df.shape)

print("\nJumlah missing values:")
display(df.isnull().sum())

print("\nJumlah baris duplikat:")
print(df.duplicated().sum())

print("\nTipe data:")
display(df.dtypes)
```

**2.1 Penanganan Data Duplikat**

Menghitung duplikat sebelum dibersihkan

```python
duplikat_sebelum = df.duplicated().sum()
```

Menghapus baris duplikat

```python
df = df.drop_duplicates().copy()
```

Memeriksa hasil

```python
print("Duplikat sebelum:", duplikat_sebelum)
print("Duplikat sesudah:", df.duplicated().sum())
print("Ukuran data sesudah:", df.shape)
```

**2.2 Pemeriksaan Missing Values**

Memeriksa nilai kosong setelah penghapusan duplikat

```python
print("Jumlah missing values setelah penghapusan duplikat:")
display(df.isnull().sum())

print("\nPersentase missing values (%):")
display((df.isnull().mean() * 100).round(2))
```

**2.3 Penanganan Missing Values**

Menangani Description yang kosong

```python
df["Description"] = df["Description"].fillna("Unknown")
```

CustomerID tidak diisi dengan nilai buatan
karena merupakan identitas pelanggan


```python
print("Missing values setelah penanganan:")
display(df.isnull().sum())

print("\nUkuran dataset:", df.shape)
```

**2.4 Transformasi Tipe Data**

Mengubah InvoiceDate menjadi tipe datetime

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print("Tipe data setelah transformasi:")
display(df.dtypes)

print("\nJumlah InvoiceDate yang kosong:")
print(df["InvoiceDate"].isnull().sum())
```

**2.5 Pemeriksaan Nilai Tidak Wajar**

Memeriksa nilai Quantity dan UnitPrice

```python
print("Quantity negatif:", (df["Quantity"] < 0).sum())
print("Quantity nol:", (df["Quantity"] == 0).sum())

print("UnitPrice negatif:", (df["UnitPrice"] < 0).sum())
print("UnitPrice nol:", (df["UnitPrice"] == 0).sum())

print("\nContoh transaksi dengan Quantity negatif:")
display(df[df["Quantity"] < 0].head())

print("\nContoh transaksi dengan UnitPrice negatif:")
display(df[df["UnitPrice"] < 0])
```

**2.6 Analisis Transaksi Khusus**

Memeriksa transaksi pembatalan

```python
print(
    "Invoice dengan awalan C:",
    df["InvoiceNo"].astype(str).str.startswith("C").sum()
)
```

Memeriksa transaksi dengan harga nol

```python
print("\nContoh transaksi dengan UnitPrice nol:")
display(df[df["UnitPrice"] == 0].head(10))

#print("\nContoh transaksi dengan UnitPrice nol:")
display(df[df["UnitPrice"] == 0].head(10))

print("\nTransaksi dengan UnitPrice negatif:")
display(df[df["UnitPrice"] < 0])
```

Memeriksa deskripsi produk yang berkaitan

```python
print("\nDeskripsi pada transaksi harga negatif:")
display(
    df.loc[df["UnitPrice"] < 0, ["Description", "Quantity", "UnitPrice", "InvoiceNo"]]
)
```

**2.7 Penanganan Transaksi dengan Harga Negatif**

Menghitung transaksi dengan harga negatif

```python
harga_negatif = (df["UnitPrice"] < 0).sum()

print("Transaksi dengan harga negatif sebelum:", harga_negatif)
```

Mengecualikan harga negatif dari dataset analisis penjualan

```python
df = df[df["UnitPrice"] >= 0].copy()

print("Transaksi dengan harga negatif sesudah:", (df["UnitPrice"] < 0).sum())
print("Ukuran dataset setelah penanganan:", df.shape)
```

**3. Hasil Setelah Preprocessing (After)**

```python

print("=== AFTER PREPROCESSING ===")

print("Ukuran dataset:", df.shape)

print("\nLima baris pertama:")
display(df.head())

print("\nJumlah missing values:")
display(df.isnull().sum())

print("\nJumlah duplikat:")
print(df.duplicated().sum())

print("\nTipe data:")
display(df.dtypes)
```

**4. Perbandingan Sebelum dan Sesudah Preprocessing**

Perbandingan kondisi sebelum dan sesudah preprocessing


```python
perbandingan = pd.DataFrame({
    "Keterangan": [
        "Jumlah baris",
        "Jumlah kolom",
        "Missing Description",
        "Missing CustomerID",
        "Jumlah duplikat",
        "Quantity negatif",
        "UnitPrice negatif",
        "UnitPrice nol"
    ],
    "Before": [
        len(online_retail.data.features),
        len(df_raw.columns),
        online_retail.data.features["Description"].isnull().sum(),
        online_retail.data.features["CustomerID"].isnull().sum(),
        df_raw.duplicated().sum(),
        (online_retail.data.features["Quantity"] < 0).sum(),
        (online_retail.data.features["UnitPrice"] < 0).sum(),
        (online_retail.data.features["UnitPrice"] == 0).sum()
    ],
    "After": [
        len(df),
        len(df.columns),
        df["Description"].isnull().sum(),
        df["CustomerID"].isnull().sum(),
        df.duplicated().sum(),
        (df["Quantity"] < 0).sum(),
        (df["UnitPrice"] < 0).sum(),
        (df["UnitPrice"] == 0).sum()
    ]
})

display(perbandingan)
```

**5. Kesimpulan**

Berdasarkan proses preprocessing pada dataset Online Retail, ditemukan beberapa masalah kualitas data, yaitu nilai kosong pada kolom Description dan CustomerID, data duplikat, perbedaan tipe data, serta transaksi dengan nilai UnitPrice negatif.

Preprocessing dilakukan dengan menghapus baris duplikat, mengganti Description yang kosong dengan label "Unknown", mengubah InvoiceDate menjadi tipe datetime, dan mengecualikan dua transaksi dengan UnitPrice negatif dari dataset analisis penjualan.

Setelah preprocessing, jumlah data berkurang dari 541.909 menjadi 536.639 baris. Jumlah duplikat menjadi nol, nilai kosong pada Description berhasil ditangani, dan tipe data tanggal berhasil diperbaiki. Sementara itu, nilai kosong pada CustomerID dan Quantity negatif tetap dipertahankan karena memerlukan pertimbangan lebih lanjut berdasarkan tujuan analisis.

Dengan demikian, data telah melalui tahap pembersihan dan transformasi awal sehingga lebih siap digunakan untuk analisis data mining berikutnya.
